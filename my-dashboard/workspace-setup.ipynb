# New Workspace Setup Guide

This notebook will guide you through setting up a new workspace for a Python project, including directory creation, Git initialization, virtual environment setup, package installation, and creating initial project files.

## 1. Create a New Workspace Directory

Use Python's `os` and `pathlib` modules to create a new directory for your workspace.

```python
import os
from pathlib import Path

# Define the workspace directory name
workspace_dir = Path('my_python_workspace')

# Create the directory if it doesn't exist
workspace_dir.mkdir(exist_ok=True)
print(f"Workspace directory created at: {workspace_dir.resolve()}")
```

## 2. Initialize a Git Repository

Run shell commands from Python to initialize a new Git repository in the workspace directory.

```python
import subprocess

# Initialize a git repository in the workspace directory
subprocess.run(['git', 'init'], cwd=workspace_dir)
print(f"Initialized empty Git repository in {workspace_dir.resolve()}/.git/")
```

## 3. Create a Python Virtual Environment

Use the `venv` module to create a virtual environment for the project.

```python
import venv

venv_dir = workspace_dir / 'venv'
venv.create(venv_dir, with_pip=True)
print(f"Virtual environment created at: {venv_dir}")
```

## 4. Install Essential Packages

Use pip to install essential packages such as numpy, pandas, and pytest in the virtual environment.

```python
# Path to pip inside the virtual environment
pip_path = venv_dir / 'bin' / 'pip'

# List of essential packages
packages = ['numpy', 'pandas', 'pytest']

# Install packages
subprocess.run([str(pip_path), 'install'] + packages)
print(f"Installed packages: {', '.join(packages)}")
```

## 5. Create Initial Project Files

Create basic files like `README.md`, `.gitignore`, and `main.py` using Python file operations.

```python
# Create README.md
readme_path = workspace_dir / 'README.md'
readme_path.write_text("""# My Python Workspace\n\nThis is the initial setup for your Python project workspace.\n""")

# Create .gitignore
ignore_path = workspace_dir / '.gitignore'
ignore_path.write_text("""venv/\n__pycache__/\n*.pyc\n""")

# Create main.py
main_path = workspace_dir / 'main.py'
main_path.write_text("""def main():\n    print('Hello, World!')\n\nif __name__ == '__main__':\n    main()\n""")

print("Initial project files created: README.md, .gitignore, main.py")
```
